In an **Initial Value Problem (IVP)**, all supplementary conditions (such as position and velocity) are specified at a single starting point $t_0$. In a **Boundary Value Problem (BVP)**, conditions are instead prescribed at two or more separate points on the boundary of the domain.

For example, consider a second-order ordinary differential equation on $[0, 1]$ with Dirichlet boundary values fixed at both endpoints:

$$\frac{\mathrm{d}^2 f}{\mathrm{d}x^2} + \frac{\mathrm{d}f}{\mathrm{d}x} = 1, \qquad f(0) = 1000, \qquad f(1) = 1$$

Because we do not know $f'(0)$ upfront, we cannot simply march forward from $x = 0$ without additional strategy. In this chapter, we study four foundational approaches for solving BVPs numerically:

1. **Shooting Method**: Guess the missing initial condition(s), integrate as an IVP, and adjust the guess via root-finding until the far boundary condition is met.
2. **Collocation Methods**: Approximate the solution as a linear combination of basis functions (such as cubic splines or radial basis functions) that satisfy the ODE at chosen collocation points.
3. **Finite Difference Method (FDM)**: Discretize the domain into a grid and replace derivatives with algebraic difference approximations.
4. **Finite Volume Method (FVM)**: Integrate conservation laws over discrete control volumes so that physical fluxes balance across cell interfaces.

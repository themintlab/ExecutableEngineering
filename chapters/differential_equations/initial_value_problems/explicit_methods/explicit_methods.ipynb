Explicit solver methods take information up to the current solution $y_i$ in order to directly predict the next step $y_{i+1}$.

## Runge-Kutta Methods

The Runge-Kutta methods are a class of time-stepping techniques where the next time step is calculated from the *current time step* and an estimate of its *slope* (rate of change). Increased accuracy is achieved through approximating higher-order derivatives to improve our time step. Since we don't generally have analytical higher derivatives, we sample the rate function $f(t, y)$ at different positions across the step to approximate them.

Since we are time stepping, our independent variable is $t$, and discretizing in time generates discrete times $t_i$ and their corresponding solutions $y(t=t_i) = y_i$.

For first-order (in time) equations, which include parabolic PDEs:

$$
\frac{\mathrm{d}y}{\mathrm{d}t} = y'_i = f(t_i, y_i)
$$

Consider the Taylor expansion:

$$
\begin{aligned}
y_{i+1} &= y_i + y'_i h + \frac{y''_i}{2}h^2 + \frac{y'''_i}{6}h^3 + \cdots \\
&= y_i + f(t_i,y_i) h + \frac{f'(t_i,y_i)}{2}h^2 + \frac{f''(t_i,y_i)}{6}h^3 + \cdots \\
&\approx y_i + h \sum_{n=1}^s a_n k_n
\end{aligned}
$$

where the last line is the general form of the Runge-Kutta methods. The weights $a_n$ are constants and $k_n$ are the slope function evaluated at different positions in the interval $[t_i, t_{i+1}]$. The goal is to match:

$$
\sum_{n=1}^s a_n k_n \approx f(t_i,y_i) + \frac{f'(t_i,y_i)}{2}h + \frac{f''(t_i,y_i)}{6}h^2 + \cdots
$$

as closely as possible with a series truncated in $s$ terms (called *stages*). Let's build up to the general form step by step.

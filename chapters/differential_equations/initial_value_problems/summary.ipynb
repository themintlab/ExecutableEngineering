**Explicit methods**:
- Are simple and inexpensive to evaluate per time step.
- Parallelize naturally across large systems.
- Suffer from conditional numerical stability.
- Require prohibitively small step sizes on **stiff** equations.

**Implicit methods**:
- Are computationally intensive per step (requiring nonlinear root finding or linear system solves).
- Do not parallelize as easily.
- Offer superior numerical stability (often unconditionally stable).
- Can take *significantly* larger step sizes on stiff problems without diverging.

## Key Takeaways Across IVP Families

- **Systems of equations** and **higher-order ODEs** (via **reduction of order**) are natural, direct vector extensions of single-equation solvers.
- **Adaptive time stepping** is essential in practice and can be achieved with minimal overhead using embedded Runge-Kutta pairs (such as **RKF45**).
- **Explicit Runge-Kutta** methods efficiently deliver high-order accuracy whenever numerical stability is not the limiting factor.
- **Implicit Runge-Kutta** methods require solving coupled stage equations simultaneously, which increases computational cost with the number of stages.
- **Explicit linear multistep** methods have step-size and stability restrictions that limit their standalone use.
- **Implicit linear multistep (Backward Differentiation Formulae, BDF)** methods combine strong stiff stability with efficient reuse of solution history.

## Dr. Mike's Tips

- **Know your physics.** If you do not know whether your system is stiff, start with an adaptive **BDF** solver (or an automatic stiffness-detecting solver).
- **Non-stiff smooth dynamics:** If you know your system is non-stiff, **RK45** is the go-to workhorse and will save substantial compute time.
- **Discontinuous forcing:** If your physics features abrupt changes (e.g., step functions, switching, or sharp pulses), consider **Backward Euler (BDF1)** across the discontinuity, where higher-order Taylor assumptions break down.

The Finite Element Method (FEM) is a very popular method with a wide range of applications. It is an *integral* technique, similar to the Finite Volume Method, which provides it with a number of benefits:

* Complex geometries
* Irregular meshing
* Irregular material properties
* Options to tweak the method according to the physics
* (quasi)dynamic mesh refinement
* Highly flexible boundary conditions
* Ability to solve *boundary physics*
* Excellent accuracy

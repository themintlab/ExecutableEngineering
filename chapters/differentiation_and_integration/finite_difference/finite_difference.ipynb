## Approach from the secant

Numerical differentiation approximates the continuous derivative of a function using discrete function evaluations. Geometrically, the true tangent slope $f'(x_0)$ is the limiting slope of secant lines connecting $(x_0, f(x_0))$ to $(x_0 + h, f(x_0 + h))$ as the step size $h$ approaches zero:

$$ f'(x_0) = \lim_{h \to 0} \frac{f(x_0 + h) - f(x_0)}{h} $$

As the step size $h$ contracts toward zero, the secant line rotates continuously into the true analytical tangent line.

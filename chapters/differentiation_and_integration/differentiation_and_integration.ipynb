In continuum mechanics, dynamics, electromagnetism, and thermal transport, physical laws are formulated through differential and integral equations.  Differentiation reveals local, instantaneous rates of change from positions or states. Integration accumulates rates, flux densities, and source distributions to yield net displacements, total charge, or mechanical work. In physical telemetry, both operations are indispensable.

## Known equations

If we knew the type of equation (or had it symbolically), we could find the derivative / integral analytically. 

> This can be extended to function comoposition via the chain rule which is the basis of backwrds differentiation in computational graphs.

Of all the families of functions, polynomials are trivial to differentiate / integrate. Fortunatley, we can always approximate the local function behaviour via a taylor expansion which will be a key tool in deriving approaches and approximating the error. 

$$
\begin{aligned}
f(x + h) &= f(x) + h f'(x) + \frac{h^2}{2!} f''(x) + \frac{h^3}{3!} f'''(x) + O(h^4) \\
f(x - h) &= f(x) - h f'(x) + \frac{h^2}{2!} f''(x) - \frac{h^3}{3!} f'''(x) + O(h^4)
\end{aligned}
$$

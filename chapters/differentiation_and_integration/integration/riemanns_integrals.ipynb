## Left and Right Riemann Integrals

Riemann''s integral is the simplest of the Newton-Cotes formulae with a polynomial of degree $0$, i.e. a constant over each subdivision. Variants exist depending on which point in the subdomain $[x_i, x_{i+1}]$ is chosen to evaluate the constant.

If we take the function value at the left of the subdomain $[x_i, x_{i+1}]$, we obtain the *left* Riemann integral:

$$\int_a^b f(x) \,\mathrm{d}x \approx \sum_{i=0}^{n-1} f(x_i) h$$

Conversely, if we take the function value at the right limit, we obtain the *right* Riemann integral:

$$\int_a^b f(x) \,\mathrm{d}x \approx \sum_{i=1}^{n} f(x_i) h$$

## Error Analysis of One-Sided Riemann Sums

To analyze the error, consider integrating the Taylor expansion of the left integral (the right being trivially similar):

$$
\begin{aligned}
f(x) &= f(x_i) + f'(x_i) h + O(h^2) \\
\int_{x_i}^{x_{i+1}} f(x) \,\mathrm{d}x &= \int_{x_i}^{x_{i+1}} [f(x_i) + f'(x_i) h + O(h^2)] \,\mathrm{d}x \\
&= f(x_i) h + f'(x_i) \frac{h^2}{2} + O(h^3) \\
&= f(x_i) h + O(h^2)
\end{aligned}
$$

The truncation error per element is $O(h^2)$. Since there are $n = (b-a)/h$ elements, the global composite error scales as:

$$E_{composite} = \sum_{i=0}^{n-1} O(h^2) = n \cdot O(h^2) = O(h)$$

making left and right Riemann sums first-order accurate.

## The Midpoint Riemann Integral

Rather than favor the left or right subdomain limit, let''s take the midpoint $y_i = \frac{x_{i+1} + x_i}{2}$. The Midpoint Rule states:

$$\int_a^b f(x) \,\mathrm{d}x \approx \sum_{i=0}^{n-1} f(y_i) h$$

## Error Analysis: Second-Order Midpoint Accuracy

The error can be deduced considering the Taylor series of $f(x)$ around the midpoint $y_i$:

$$
\begin{aligned}
f(x) &= f(y_i) + f'(y_i)(x - y_i) + \frac{f''(y_i)}{2}(x - y_i)^2 + O(h^3) \\
\int_{x_i}^{x_{i+1}} f(x) \,\mathrm{d}x &= \int_{y_i - h/2}^{y_i + h/2} \left[f(y_i) + f'(y_i)(x - y_i) + \frac{f''(y_i)}{2}(x - y_i)^2 + O(h^3)\right] \,\mathrm{d}x
\end{aligned}
$$

Because the integration limits are symmetric around $y_i$, the odd term vanishes: $\int_{-h/2}^{h/2} (x - y_i) \,\mathrm{d}x = 0$.

Integrating the quadratic term yields:

$$\int_{x_i}^{x_{i+1}} f(x) \,\mathrm{d}x = f(y_i) h + \frac{f''(y_i)}{24} h^3 + O(h^5)$$

Summing across all $n$ subdivisions gives a composite error of $O(h^2)$, a full order of accuracy higher than one-sided Riemann sums!

*Note*: If you only have discrete data measured at points $x_i$, you cannot evaluate $f(y_i)$ without interpolating between samples.
